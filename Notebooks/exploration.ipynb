```python
import mne

raw = mne.io.read_raw_edf(r"C:\Users\Admin\Documents\Thesis\FileSenderDownload_28-2-2026__19-22-42\EDF-extracties\EDF-extracties\3.edf", preload=True)

raw.plot(duration=5, n_channels=10)
```

```python
from scipy.signal import spectrogram
import matplotlib.pyplot as plt

data = raw.get_data()

signal = data[0]  # first channel

fs = raw.info['sfreq']

f, t, Sxx = spectrogram(signal, fs)

plt.pcolormesh(t, f, Sxx)
plt.ylabel('Frequency (Hz)')
plt.xlabel('Time (s)')
plt.title('Spectrogram of EEG Signal')
plt.show()
```

```python

import mne

file_path = r"C:\Users\Admin\Documents\Thesis\FileSenderDownload_28-2-2026__19-22-42\EDF-extracties\EDF-extracties\3.edf"

raw = mne.io.read_raw_edf(file_path, preload=True)

print(raw)
```

```python
start_time = 5   # skip first 5 seconds (safe)
duration = 10    # show 10 seconds window
```

```python
raw.plot(
    start=start_time,
    duration=duration,
    n_channels=10,
    scalings='auto',
    show=True
)
```

```python
fig = raw.plot(
    start=start_time,
    duration=duration,
    n_channels=10,
    scalings='auto',
    show=False,
    block=False
)

for ax in fig.axes:
    ax.set_title("")  # cleaner look

fig.savefig("clean_eeg.png", dpi=300, bbox_inches='tight')
```
